```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
from google.colab.patches import cv2_imshow
from PIL import Image
import tensorflow as tf
from tensorflow import keras
from keras.datasets import mnist
from tensorflow.math import confusion_matrix

# Load the MNIST dataset
(X_train, Y_train), (X_test, Y_test) = mnist.load_data()
```

```python
# Step 3: Understand Dataset Shape
print(f"Training Data Shape: {X_train.shape}, Labels: {Y_train.shape}")

# Step 4: Visualize Sample Image (10th image)
print("Visualizing the 10th training image:")
plt.imshow(X_train[10], cmap='gray')
plt.show()

# Step 5: Normalize the Dataset
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
# Step 6: Build Neural Network Model
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(50, activation='relu'),
    keras.layers.Dense(50, activation='relu'),
    keras.layers.Dense(10, activation='sigmoid') # Using the structure from your docs
])

# Step 7: Compile the Model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Step 8: Train the Model (Internship doc suggests 10 epochs)
model.fit(X_train, Y_train, epochs=10)
```

```python
# Step 9: Evaluate the Model
loss, accuracy = model.evaluate(X_test, Y_test)
print(f'Test Accuracy: {accuracy*100:.2f}%')

# Step 10 & 11: Make Predictions and use Argmax
Y_pred = model.predict(X_test)
# Look at the first image in test set
predicted_label = np.argmax(Y_pred[0])

# Step 12: Display Prediction Result
print(f"Predicted Label for first test image: {predicted_label}")
print(f"Actual Label for first test image: {Y_test[0]}")
```
